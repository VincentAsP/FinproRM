```python
import cv2
import os
import glob

# 1. Atur folder asal (tempat banyak video) dan folder tujuan (tempat gambar hasil)
folder_video = '../FixedDataset/Not' 
folder_tujuan = 'train_dataset/negative'

# Buat folder tujuan jika belum ada
if not os.path.exists(folder_tujuan):
    os.makedirs(folder_tujuan)

# Cari semua file video (.mp4) di dalam folder tersebut
daftar_video = glob.glob(os.path.join(folder_video, '*.mp4'))

frame_tersimpan = 0
ambil_setiap_frame = 15 # Ambil 1 gambar setiap 15 frame

print(f"Ditemukan {len(daftar_video)} video. Mulai mengekstrak...")

# 2. Mesin akan looping membuka video satu per satu
for video_path in daftar_video:
    print(f"Memproses: {os.path.basename(video_path)}")
    cap = cv2.VideoCapture(video_path)
    count = 0
    
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break # Video ini habis, lanjut ke video berikutnya
            
        if count % ambil_setiap_frame == 0:
            # Format nama file dibuat berurutan agar tidak tertimpa
            nama_file = f"{folder_tujuan}/frame_{frame_tersimpan:04d}.jpg"
            cv2.imwrite(nama_file, frame)
            frame_tersimpan += 1
            
        count += 1
    cap.release()

print(f"Beres! Total dapet {frame_tersimpan} gambar dari seluruh video.")
```
